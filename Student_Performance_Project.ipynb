<h2 id="%F0%9F%94%8E-Phase-3-%E2%80%94-Data-Analysis">🔎 Phase 3 — Data Analysis</h2><p>We now compare average subject scores across different student groups to explore possible relationships with performance.</p>

```python
# --------------------------------------------------------------
#                 ➗ MATHEMATICS SCORE ANALYSIS
# --------------------------------------------------------------

# Compare average mathematics scores by test preparation status
# groupby() separates students by preparation status before calculating the mean.

df.groupby('test preparation course')['math score'].mean()
```

```python
# --------------------------------------------------------------
#                 📖 READING SCORE ANALYSIS
# --------------------------------------------------------------

# Compare average reading scores by test preparation status

df.groupby('test preparation course')['reading score'].mean()
```

```python
# --------------------------------------------------------------
#                 ✍️ WRITING SCORE ANALYSIS
# --------------------------------------------------------------

# Compare average writing scores by test preparation status

df.groupby('test preparation course')['writing score'].mean()
```

```python
# --------------------------------------------------------------
#                 📊 SUBJECT-WISE COMPARISON
# --------------------------------------------------------------

# Display all three subject averages together
# A single table makes it easier to compare the groups across subjects.

df.groupby('test preparation course')[['math score', 'reading score', 'writing score']].mean()
```

```python
# --------------------------------------------------------------
#                 🎓 PARENTAL EDUCATION COMPARISON
# --------------------------------------------------------------

# Compare average subject scores across parental education levels
# This helps us examine how the averages differ between education categories.

df.groupby('parental level of education')[['math score', 'reading score', 'writing score']].mean()
```

```python
# --------------------------------------------------------------
#                 👥 GENDER COMPARISON
# --------------------------------------------------------------

# Compare average subject scores by gender
# The result shows the average performance in each subject for each gender group.

df.groupby('gender')[['math score', 'reading score', 'writing score']].mean()
```

```python
# --------------------------------------------------------------
#                 📈 OVERALL SUBJECT AVERAGES
# --------------------------------------------------------------

# Calculate the overall average for each subject
# This gives us a general performance benchmark for the full dataset.

df[['math score', 'reading score', 'writing score']].mean()
```

```python
# --------------------------------------------------------------
#                 🔗 SUBJECT SCORE CORRELATION
# --------------------------------------------------------------

# Examine the correlation between the three subject scores
# Correlation values help show how strongly two sets of scores move together.

df[['math score', 'reading score', 'writing score']].corr()
```

```python
# --------------------------------------------------------------
#                 🔥 CORRELATION HEATMAP
# --------------------------------------------------------------

# Visualise the relationship between Mathematics, Reading and Writing scores.
# Values closer to 1 indicate a stronger positive relationship.

corr_matrix = df[['math score', 'reading score', 'writing score']].corr()

plt.figure(figsize=(7, 5))
plt.imshow(corr_matrix, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar(label='Correlation')
plt.xticks(range(len(corr_matrix.columns)), corr_matrix.columns, rotation=20)
plt.yticks(range(len(corr_matrix.columns)), corr_matrix.columns)
plt.title('Correlation Between Subject Scores')

for i in range(len(corr_matrix.columns)):
    for j in range(len(corr_matrix.columns)):
        plt.text(j, i, round(corr_matrix.iloc[i, j], 3),
                 ha='center', va='center')

plt.tight_layout()
plt.show()
```

<h2 id="%F0%9F%93%88-Phase-4-%E2%80%94-Visualisation-and-Insights">📈 Phase 4 — Visualisation and Insights</h2><p>Charts make patterns easier to notice. We use Matplotlib and pandas plotting to visualise score distributions and group comparisons.</p>

```python
# --------------------------------------------------------------
#                 🎨 PHASE 4: VISUALISATION & INSIGHTS
# --------------------------------------------------------------

# Import Matplotlib for creating visualisations

import matplotlib.pyplot as plt
```

```python
# --------------------------------------------------------------
#                 📊 MATHEMATICS SCORE DISTRIBUTION
# --------------------------------------------------------------

# Visualise the distribution of mathematics scores
# The histogram groups scores into intervals and shows the number of students in each interval.

plt.hist(df['math score'], bins=10)
plt.title('Distribution of Mathematics Scores')
plt.xlabel('Math Score')
plt.ylabel('Number of Students')
plt.show()
```

```python
# --------------------------------------------------------------
#                 📊 READING SCORE DISTRIBUTION
# --------------------------------------------------------------

# Visualise the distribution of reading scores

plt.hist(df['reading score'], bins=10)
plt.title('Distribution of Reading Scores')
plt.xlabel('Reading Score')
plt.ylabel('Number of Students')
plt.show()
```

```python
# --------------------------------------------------------------
#                 📊 WRITING SCORE DISTRIBUTION
# --------------------------------------------------------------

# Visualise the distribution of writing scores

plt.hist(df['writing score'], bins=10)
plt.title('Distribution of Writing Scores')
plt.xlabel('Writing Score')
plt.ylabel('Number of Students')
plt.show()
```

```python
# --------------------------------------------------------------
#                 📊 TEST PREPARATION COMPARISON
# --------------------------------------------------------------

# Compare subject averages for students with and without test preparation
# First calculate the group averages, then display them as a bar chart.

prep_scores = df.groupby('test preparation course')[['math score', 'reading score', 'writing score']].mean()

prep_scores.plot(kind='bar')
plt.title('Average Scores by Test Preparation')
plt.xlabel('Test Preparation Course')
plt.ylabel('Average Score')
plt.xticks(rotation=0)
plt.show()
```

```python
# --------------------------------------------------------------
#                 🎓 PARENTAL EDUCATION COMPARISON
# --------------------------------------------------------------

# Compare average subject scores across parental education levels
# The bar chart makes it easier to compare the different education categories.

parent_scores = df.groupby('parental level of education')[['math score', 'reading score', 'writing score']].mean()

parent_scores.plot(kind='bar')
plt.title('Average Scores by Parental Education')
plt.xlabel('Parental Level of Education')
plt.ylabel('Average Score')
plt.xticks(rotation=45)
plt.show()
```

```python
# --------------------------------------------------------------
#                 👥 GENDER COMPARISON
# --------------------------------------------------------------

# Compare average subject scores by gender

gender_scores = df.groupby('gender')[['math score', 'reading score', 'writing score']].mean()

gender_scores.plot(kind='bar')
plt.title('Average Scores by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Score')
plt.xticks(rotation=0)
plt.show()
```

```python
# --------------------------------------------------------------
#                 📦 SUBJECT SCORE BOXPLOT
# --------------------------------------------------------------

# Use a boxplot to compare the spread of scores across subjects
# The plot helps us compare medians, variation, and possible unusual values.

df[['math score', 'reading score', 'writing score']].plot(kind='box')

plt.title('Comparison of Student Scores')
plt.ylabel('Scores')
plt.show()
```

<h2 id="%F0%9F%A7%A0-Phase-5-%E2%80%94-Analytical-Evaluation">🧠 Phase 5 — Analytical Evaluation</h2><p>In this phase, we summarise key score comparisons and calculate an overall score for each student.</p>

```python
# --------------------------------------------------------------
#                 🏆 PHASE 5: ANALYTICAL EVALUATION
# --------------------------------------------------------------

# Identify the subject with the highest and lowest average score
# idxmax() and idxmin() return the subject names for the largest and smallest averages.

subject_avg = df[['math score', 'reading score', 'writing score']].mean()

print("Highest Average Score:")
print(subject_avg.idxmax(), subject_avg.max())

print("Lowest Average Score:")
print(subject_avg.idxmin(), subject_avg.min())
```

```python
# --------------------------------------------------------------
#                 ➕ CALCULATING TOTAL SCORE
# --------------------------------------------------------------

# Create a total score by adding the three subject scores
# This new column gives one combined score for each student.

df['total score'] = df['math score'] + df['reading score'] + df['writing score']

# Calculate the average total score across all students.
df['total score'].mean()
```

```python
# --------------------------------------------------------------
#                 🥇 HIGHEST TOTAL SCORE
# --------------------------------------------------------------

# Find the highest total score achieved in the dataset

df['total score'].max()
```

```python
# --------------------------------------------------------------
#                 📉 LOWEST TOTAL SCORE
# --------------------------------------------------------------

# Find the lowest total score recorded in the dataset

df['total score'].min()
```

```python
# --------------------------------------------------------------
#                 📊 TOTAL SCORE BY TEST PREPARATION
# --------------------------------------------------------------

# Compare average total scores by test preparation status
# This provides an overall comparison of the two preparation groups.

df.groupby('test preparation course')['total score'].mean()
```

```python
# --------------------------------------------------------------
#                 💾 EXPORTING THE FINAL DATASET
# --------------------------------------------------------------

# Save the updated dataset as a CSV file
# index=False prevents pandas from adding an extra index column to the exported file.

df.to_csv('Student Performance Analysis.csv', index=False)

# Download the final CSV file from Google Colab.
files.download('Student Performance Analysis.csv')
```
